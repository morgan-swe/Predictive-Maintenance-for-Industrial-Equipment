# Machine Learning Models

## Baseline Linear Regression

Objective

Train a baseline regression model for Remaining Useful Life prediction.

## Load libraries and dataset

Load the prepared dataset

```python
import pandas as pd
df = pd.read_csv('../data/processed/train_FD001_features.csv')

df.info()
```

```python
import os
os.makedirs("../reports/figures", exist_ok=True)
```

# Recreate X and y

The target is RUL, everything else other than identifiers like `unit_number` becomes our features.

## Split the data

```python
from sklearn.model_selection import GroupShuffleSplit

# 0. Drop warm up rows and leaky columns
df_model = df[df['time_in_cycles'] >= 5].reset_index(drop=True)

groups = df_model['unit_number']
x = df_model.drop(columns=['RUL', 'max_cycles', 'unit_number'])

# Cap RUL at 125 cycles.
RUL_CAP = 125
y = df_model['RUL'].clip(upper=RUL_CAP)

# 1. Split by engine; 80 engines train, 20 engines test
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(x, y, groups=groups))

x_train, x_test = x.iloc[train_idx], x.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print(x_train.shape, x_test.shape)
print("train engines:", groups.iloc[train_idx].nunique(),
      "test engines:", groups.iloc[test_idx].nunique())
print("overlap:", set(groups.iloc[train_idx]) & set(groups.iloc[test_idx]))
```

## Import linear regression

Initialize and fit the model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

## Predict values for unseen test data

```python
lr_predictions = model.predict(x_test)
```

## Evaluate linear regression model

Using three metrics:
- Mean absolute error
- Root mean squared error
- R² Score

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

model_results = []

def evaluate(name, y_true, y_hat):
    """Score a model, record the result, and print it."""
    scores = {
        "Model": name,
        "MAE": mean_absolute_error(y_true, y_hat),
        "RMSE": root_mean_squared_error(y_true, y_hat),
        "R2": r2_score(y_true, y_hat),
    }

    # Overwrite any existing row for this model so re-running a cell updates it
    # in place instead of appending a duplicate.
    for i, row in enumerate(model_results):
        if row["Model"] == name:
            model_results[i] = scores
            break
    else:
        model_results.append(scores)

    print(name)
    print(f"MAE:  {scores['MAE']:.2f} cycles")
    print(f"RMSE: {scores['RMSE']:.2f} cycles")
    print(f"R²:   {scores['R2']:.3f}")

    return scores
```

```python
evaluate("Linear Regression", y_test, lr_predictions)
```

# Plot actual vs predictions - Linear regression

Compare the predicted values to the actual values using scatter plots.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(y_test, lr_predictions, alpha=0.2, s=8, edgecolor='none')

# Perfect prediction reference line
lims = [min(y_test.min(), lr_predictions.min()), max(y_test.max(), lr_predictions.max())]
ax.plot(lims, lims, 'r--', linewidth=1.5, label='Perfect prediction')

ax.axhline(0, color='gray', linewidth=0.8, linestyle=':')

ax.set_xlabel('Actual RUL (cycles)')
ax.set_ylabel('Predicted RUL (cycles)')
ax.set_title('Predicted vs Actual RUL — Linear Regression (held-out engines)')
ax.set_aspect('equal')
ax.legend()

# Save the plot
plt.tight_layout()

plt.savefig("../reports/figures/prediction_scatter.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

# Linear regression observations

- Points below the red line = the model predicts less life than the engine actually had.
- Points above the line = the model predicts more life than there was. These are the dangerous ones: the engine fails sooner than you were told.
- Predictions below zero. Linear regression is unbounded, so it will output negative RUL.


# Residual plot - Linear regression

Residuals give us a sense of the overall quality, or accuracy of the predictions made with the line.

```python
lr_residuals = y_test - lr_predictions

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(lr_predictions, lr_residuals, alpha=0.2, s=8, edgecolor='none')
ax.axhline(0, color='red', linestyle='--', linewidth=1.5)
ax.set_xlabel('Actual RUL (cycles)')
ax.set_ylabel('Residual (actual − predicted)')
ax.set_title('Residuals vs Actual RUL — Linear Regression')

# Save the plot
plt.tight_layout()

plt.savefig("../reports/figures/linear_regressionresidual_plot.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

# Decision tree regression

The relationship between wear & tear vs health is often non linear. Decision trees can capture these non linear relationships.

```python
from sklearn.tree import DecisionTreeRegressor

# An unconstrained tree grows until every leaf is pure, which memorises noise
# across 88 correlated features. Limiting depth and leaf size keeps it
# generalising; random_state makes the result reproducible.
decision_tree = DecisionTreeRegressor(
    max_depth=6,
    min_samples_leaf=20,
    random_state=42,
)
decision_tree.fit(x_train, y_train)

# A large train/test gap here is the signature of overfitting.
print(f"train R²: {decision_tree.score(x_train, y_train):.3f}")
print(f"test R²:  {decision_tree.score(x_test, y_test):.3f}")
```

```python
dt_predictions = decision_tree.predict(x_test)
```

```python
evaluate('Decision Tree', y_test, dt_predictions)
```

# Decision tree observations

- Decision trees don't fit an equation. It carves the feature space into boxes with yes/no splits, and every row that lands in the same box gets the same prediction, hence the horizontal bands we see inside the plot.
- A constraint was added capping `max_depth=6` making the tree 2⁶=64 leaves, so the predictions take at most 64 distinct values across 3990 test rows.

# Compare feature types

The per-feature ranking answers "which single feature mattered most". This asks
a different question: **which kind of feature engineering paid off**.

Group all 88 features into four categories and compare the importance each one
earned.

```python
def feature_type(name):
    """Sort a feature name into one of the four engineering categories."""
    if name.endswith("_roll_mean_5"):
        return "Rolling mean"
    if name.endswith("_roll_std_5"):
        return "Rolling std"
    if name.endswith("_change"):
        return "Trend (change)"
    if name.startswith("sensor_measurement_"):
        return "Raw sensor"
    return "Other (time / settings)"

importance_df["Type"] = importance_df["Feature"].apply(feature_type)

type_summary = (
    importance_df
    .groupby("Type")["Importance"]
    .agg(Features="count", Total="sum", Average="mean")
    .sort_values("Average", ascending=False)
)

type_summary
```

# Key findings

## For feature importance

- Trend and roll standard deviation features flopped. Both features failed for the same reason: they are tools for detecting a kind of change that this dataset doesn't contain. The degradation in C-MAPSS is a slow, steady drift.
- Trend failed because every reading the change is tiny, the numbers mostly leave noise and barely any signal.
- Rolling std failed because it measure how much readings bounce around, however in our case the numbers just drift around. Failing parts can get erratic before their average shifts, something like this would genuinely hold for vibration data which just isn't the failure mode simulated in this dataset.
- Rolling mean won because averaging cancels out noise.
- A feature isn't good or bad in the abstract. It's a question you're asking the data, and it only pays off if that question matches the real failure mechanism.

# Hyperparameter tuning observations

- Hyperparameter tuning tests different paramteres for better results.
- GridSearchCV systematically searches every parameter combination using cross validation inside the training data, so we get a lower noise comparison and doesnt't contaminate the test set by tuning against it.
- Hyperparameter tuning resulted in a modest improvement in Mean Absolute Error (10.38 to 10.23 cycles). However, RMSE and R² remained nearly unchanged. This suggests that the baseline Random Forest was already well configured for the FD001 dataset, and further tuning yielded only marginal gains.
- Random forests are already self regularizing, averaging many decorrelated trees does most of the work that tuning would otherwise do, and our signal is concentrated in a couple of strong smoothed features, so the model saturates near its ceiling almost regardless of the settings.

```python
error_df = pd.DataFrame({
    # Keep the identifiers: without them a row of errors is anonymous and you
    # can't tell one bad engine from a genuinely weak model.
    "unit_number": df_model["unit_number"].loc[y_test.index],
    "time_in_cycles": df_model["time_in_cycles"].loc[y_test.index],
    "Actual_RUL": y_test,
    "Predicted_RUL": rf_predictions
})
```

```python
error_df["Error"] = (
    error_df["Actual_RUL"]
    - error_df["Predicted_RUL"]
)

error_df["Absolute_Error"] = (
    error_df["Error"].abs()
)
```

```python
worst_predictions = error_df.sort_values(
    by="Absolute_Error",
    ascending=False
)

worst_predictions.head(10)
```

# Worst prediction observations

- The ten largest errors all belong to a single engine, unit 84. Each engine contributes ~200 consecutive rows whose errors move together, so the top of a row-level ranking will always be one engine's worst streak. Rank engines, not rows.
- The errors run in the opposite direction to what the table first suggests. `Actual_RUL` in those rows is 108-125, the *highest* values in the data; it's `Predicted_RUL` that is small (43-68). The model is calling nearly-new engines half-dead, not the other way round.
- Accuracy is best at low RUL, not high. Grouped by actual RUL, MAE is 3.9 cycles in the 0-25 band, peaks at 18.1 in the 51-75 band, and falls back to 6.6 on the capped rows. This is the result we want: the model is sharpest in the window where a maintenance decision actually gets made.
- Direction matters more than magnitude. Under-prediction (unit 84) means servicing an engine early, which is wasteful but safe. Over-prediction (units 45, 74, 77) promises life that isn't there. Absolute error treats these as equal; the maintenance schedule does not.
- Part of the error is an artifact of the RUL cap. 1490 of 3990 test rows sit at 125 and account for 24% of total absolute error. A healthy engine at cycle 20 and one at cycle 90 both get labelled 125 but look different to the sensors, so the model is penalised for a distinction the label erased.
