# Load libraries and dataset

```python
import pandas as pd
train_df = pd.read_csv('../data/interim/train_FD001.csv')

max_engine_cycles = (
    train_df
        .groupby("unit_number")["time_in_cycles"]
        .max()
)
max_engine_cycles = max_engine_cycles.reset_index().rename(columns={"time_in_cycles": "max_cycles"})
```

Combine the orginal dataset with the max engine cycles one so that every row has the maximum cycle for its engine. 

```python
df_merge = pd.merge(train_df, max_engine_cycles, on="unit_number")
df_merge.describe()
```

We can create the remaining useful life column by calculating max_cycles - time_in_cycles

```python
df_merge["RUL"] = df_merge["max_cycles"] - df_merge["time_in_cycles"]
df_merge[df_merge["unit_number"] == 1]
```

# Rolling Statistics

This section creates rolling averages for important sensors to capture recent engine behavior for sensor measurement 9.

```python
df_merge['sensor_measurement_9_roll_mean_5'] = (
    df_merge
    .groupby('unit_number')['sensor_measurement_9']
    .rolling(window=5, min_periods=1)
    .mean()
    .reset_index(level=0, drop=True)
)
```

## Verifying results

```python
engine_1 = df_merge[df_merge['unit_number'] == 1]
engine_1[['time_in_cycles', 'sensor_measurement_9', 'sensor_measurement_9_roll_mean_5']]
```

# Visualizing the difference

```python
plt.plot(engine_1['time_in_cycles'], engine_1['sensor_measurement_9'], label='Raw sensor measurement')
plt.plot(engine_1['time_in_cycles'], engine_1['sensor_measurement_9_roll_mean_5'], label='Rolling mean (window=5)')
plt.title("Rolling mean vs sensor measurement for sensor 9")
plt.xlabel('Time in Cycles')
plt.ylabel('Sensor measurement 9')
plt.legend()

# Save the plot
plt.savefig("../reports/figures/sensor9_rolling_mean.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

# Rolling average observations

- The rolling average smooths the data by averaging out the short term noise and fluctuations, revealing the underlying trend in the sensor signal rather than its instantaneous value.
-  The individual, cycle-by-cycle sensor readings were lost; short-term fluctuations and any single cycle spikes are averaged away. This also means the first 4 rows per engine became NaN, since a 5-cycle window has no data to average until the 5th reading.
- The overall degradation trend for each engine became easier to see; instead of a noisy, jagged signal, the rolling mean shows a clearer directional pattern in sensor behavior as the engine ages.
- A model trained on raw sensor values can overfit to sensor noise and single cycle spikes that don't reflect real equipment condition. A rolling mean gives the model a feature that reflects sustained trend rather than momentary noise, which is more likely to generalize and correlate with actual degradation; making it a more reliable signal for predicting RUL.

# Rolling Standard Deviation

This section creates rolling standard deviation features that measure short-term variability in sensor readings for sensor measurement 9.

```python
df_merge['sensor_measurement_9_roll_std_5'] = (
    df_merge.groupby('unit_number')['sensor_measurement_9']
    .rolling(window=5)
    .std()
    .reset_index(level=0, drop=True)
)
```

## Verifying the results

```python
engine_1 = df_merge[df_merge['unit_number'] == 1]
engine_1[['time_in_cycles', 'sensor_measurement_9', 'sensor_measurement_9_roll_std_5']]
```

# Visualization

```python
fig, ax1 = plt.subplots()
ax1.plot(engine_1['time_in_cycles'], engine_1['sensor_measurement_9'], color='tab:blue', label='Raw sensor measurement')
ax1.set_xlabel('Time in Cycles')
ax1.set_ylabel('Sensor measurement 9', color='tab:blue')

ax2 = ax1.twinx()
ax2.plot(engine_1['time_in_cycles'], engine_1['sensor_measurement_9_roll_std_5'], color='tab:orange', label='Rolling std (window=5)')
ax2.set_ylabel('Rolling std', color='tab:orange')

plt.title("Rolling standard deviation vs sensor measurement for sensor 9")
fig.legend(loc='upper right')

# Save the plot
plt.savefig("../reports/figures/sensor9_rolling_std.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

# Rolling standard deviation observations

- It measures how much the sensor values fluctuate within a window, rather than where they're centered.
- A component starting to fail often becomes erratic before its average value visibly shifts with vibration, pressure, or temperature readings start swinging more before the mean drifts. A rolling average can mask this: if a sensor oscillates between higher and lower values, those can cancel out and leave the mean looking roughly normal. Rolling std picks up on that growing instability directly, since it's measuring spread.
- Vibration sensors are the classic example, a bearing or rotating component's average vibration level might barely move as it wears, but the variance/spread of vibration readings increases noticeably as imbalance, looseness, or micro fractures develop.

# Trend Features

This section creates features that measure how quickly sensor values change over time for sensor measurement 9.

```python
df_merge['sensor_measurement_9_change'] = (
    df_merge.groupby('unit_number')['sensor_measurement_9'].diff()
)
```

## Verifying the results

```python
engine_1 = df_merge[df_merge['unit_number'] == 1]
engine_1[['time_in_cycles', 'sensor_measurement_9', 'sensor_measurement_9_change']]
```

# Observations

- Trend feature measures the change in sensor readings for each cycle.
- A trend feature captures the direction and rate of change of a sensor over time. While a single sensor reading only provides the current state, the trend reveals whether the measurement is consistently increasing or decreasing, which may indicate equipment degradation before absolute values become abnormal.
- The first value is undefined because there is no previous cycle for comparison. Leaving it as NaN during feature engineering preserves the integrity of the data, and missing values can be handled later during preprocessing.

# Automating feature engineering

## Rolling mean

Add rolling mean for all sensor measurements.

```python
import re

sensor_columns = [
    col for col in df_merge.columns
    if re.fullmatch(r'sensor_measurement_\d+', col)
]

for sensor in sensor_columns: 
    df_merge[f'{sensor}_roll_mean_5'] = (
        df_merge
        .groupby('unit_number')[sensor]
        .rolling(window=5)
        .mean()
        .reset_index(level=0, drop=True)
    )
```

## Rolling standard deviation

Add rolling standard deviation for all sensor measurements.

```python
for sensor in sensor_columns: 
    df_merge[f'{sensor}_roll_std_5'] = (
        df_merge.groupby('unit_number')[sensor]
        .rolling(window=5)
        .std()
        .reset_index(level=0, drop=True)
    )
```

## Trend feature

Add trend features for all sensor measurements.

```python
for sensor in sensor_columns:
    df_merge[f'{sensor}_change'] = (
        df_merge.groupby('unit_number')[sensor].diff()
    )
```

# Observations

- Missing values come from calculating rolling mean, rolling standard deviation, and trend features, not from any of the raw sensor or operational columns, which are complete.
- Rolling mean and rolling standard deviation both have exactly 400 missing values per column, since a 5-cycle window needs 5 readings before it can produce a value. With 100 engines in the dataset, that's the first 4 rows of each engine left undefined, 100 x 4 = 400, which lines up with the count above for sensor 9 and all 21 automated rolling mean columns.
- The trend feature has exactly 100 missing values, one per engine, since it only needs the previous cycle rather than a full window; the very first cycle of each engine has nothing before it to compare against.
- This isn't random missingness, it's a direct and predictable consequence of how window and lag based features are calculated at the start of each engine's life. That means it shouldn't be filled in with something generic like a mean, since that would invent a value that was never really there.
- Next step is deciding how to handle these rows before modeling, either dropping them, relaxing the rolling window so early cycles still get a value, or leaving the NaNs for the model/pipeline to handle explicitly. I'll make that call once I can see how much early-cycle data actually matters for predicting RUL.

# Feature scaling and Data splitting

## Train/Test split data

```python
import numpy as np
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# 0. Drop warm up rows and leaky columns
df_model = df_merge[df_merge['time_in_cycles'] >= 5].reset_index(drop=True)

groups = df_model['unit_number']
x = df_model.drop(columns=['RUL', 'max_cycles', 'unit_number'])
y = df_model['RUL']

# 1. Split by engine; 80 engines train, 20 engines test
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(x, y, groups=groups))

x_train, x_test = x.iloc[train_idx], x.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# 2. Scale fit on train only
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

print(x_train.shape, x_test.shape)
print("train engines:", groups.iloc[train_idx].nunique(),
      "test engines:", groups.iloc[test_idx].nunique())
print("overlap:", set(groups.iloc[train_idx]) & set(groups.iloc[test_idx]))
```

# Observations

- We split the data before scaling so that we hide some data from the model. It learns from the other part, and then you test it on the hidden part. Without doing this you have no idea whether the model learned something real or just memorized.
- A data leakage is when information sneaks into your features that wouldn't be available in real life, so the model looks amazing in testing and falls apart in production.
- A random forest isn't needed for scaling because a tree only ever asks if a value is above or below a threshold. It never adds, multiplies, or measures distanve between features.
- If we wanted to retrain a model next week on new data we would need to fit a new scaler. Retraining a model would mean that the data is the new training set, so the transformation should describe it.
